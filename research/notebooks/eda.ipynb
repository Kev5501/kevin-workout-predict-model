# EDA

```python
import gspread
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Pull Data
import os
from dotenv import find_dotenv, load_dotenv

load_dotenv(find_dotenv(usecwd=True))  
gc = gspread.service_account(filename=os.environ["GOOGLE_CREDENTIALS_PATH"])
spreadsheet = gc.open("Hypertorphy Progress Tracker")
worksheet = spreadsheet.worksheet("Enriched_Data")
data = worksheet.get_all_records()
```

```python
#Summary
df = pd.DataFrame(data)
df.columns = df.columns.str.lower()
df.head()
```

```python
#Casting
date_cols = ['date']
float_cols = ['weight','rpe','rpe_norm_weight','body_weight']
str_cols = ['set_id','gym_day','exercise','type','comment']
int_cols = [col for col in df.columns if col not in float_cols + str_cols + date_cols]

#Date
df['date'] = pd.to_datetime(df['date']).dt.date

#Cast Ints
for col in int_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce').astype('Int64')

#Cast floats
for col in float_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce').astype(float)
```

```python
df.describe()
```

```python
#Row counts
print(f'{len(df)} Total Sets')
print(f'{len(df.groupby(['date','gym_day']))} Total Workouts')

df.groupby(['exercise']).count('date')
```

```python
#Sets done per exercise
df.groupby('exercise')[['date']].count().reset_index().sort_values(by='date', ascending = False)
```
